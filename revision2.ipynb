RECURSION

When a function calls itself repeatedly

```python
#Recursive Function
def show(n):
    if(n == 0): #base case like break in loops
        return #control return
    print(n)
    show(n-1) 
    
show(5)
```

```python
#RECURSION return n!
def fact(n):
    if (n == 1 or n == 0):
        return  1
    return fact(n-1)*n 

print(fact(3))

```

```python
#write a recursive function to calculate the sum of first n natural numbers.
def Calculate(n):
    if(n == 0):
        return 0
    return  Calculate(n-1)+n 
    
Calculate(5)
    
```

```python
#write a recursive function to print all elements in a list Hint : use list and index as parameter.
def print_list(list, idx =0):
    if (idx == len(list)):
        return 
    print(list[idx])
    print_list(list, idx+1)
fruits = ["Mango", "Strawberry", "Apple",  "Litchyee"]
print_list(fruits)
```

INHERITANCE

When one class (child/derived) drives the properties & methods of another class(parents/base).

```python
class Car:
    color = "black"
    @staticmethod
    def start():
        print("car started...!")
        
    @staticmethod
    def stop():
        print("Car stopped.")
        
class ToyotaCar(Car):
    def __init__(self, name):
        self.name = name 
        
car1 = ToyotaCar("Fortuner")
car2 = ToyotaCar("Prius")

print(car1.start())
car1.color
```

Super Method 

super() method is used to access methods of the parent class.

```python
class Car:
    def __init__(self, type):
        self.type = type 
        
    @staticmethod
    def start():
        print("car started...!")
        
    @staticmethod
    def stop():
        print("Car stopped.")
        
class ToyotaCar(Car):
    def __init__(self, name , type):
        self.name = name 
        super().__init__(type)
        
car1 = ToyotaCar("Prius", "electric")
car1.type
        
```

Property 

we use @property decorator on any method in the class to use the method as a property.


```python
class Student:
    def __init__(self, phy, chem, maths):
        self.phy = phy 
        self.chem = chem 
        self.maths = maths 
        self.percentage = str((self.phy + self.chem + self.maths)/3) + "%"     
    
        @property
        def percentage(self):
            return str((self.phy + self.chem + self.maths)/3) + "%" 
    
    
    
stu1 = Student(86, 82, 63)
print(stu1.percentage) 

stu1.phy = 74
stu1.chem = 85
print(stu1.percentage)
```

```python
class Complex:
    def __init__(self, real, img):
        self.real = real 
        self.img = img 
        
    def showNumber(self):
        print(self.real, "i +", self.img, "j") 
        
    def __add__(self, num2): #dunder fuction
        newReal = self.real + num2.real 
        newImg = self.img + num2.img 
        return Complex(newReal, newImg)
        
num1 = Complex(1,3)
num1.showNumber()

num2 = Complex(4,6)
num2.showNumber() 

num3 = num1 + num2
num3.showNumber()
```
