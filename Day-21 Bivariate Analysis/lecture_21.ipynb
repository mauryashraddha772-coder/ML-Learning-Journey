```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
tips = pd.read_csv(r"tip.csv")
tips
```

```python
titanic = pd.read_csv(r"train.csv")
titanic.head()
```

```python
iris = pd.read_csv(r"Iris-(1).csv")
iris.head()
```

# Bivariate Analysis & Multivariate

1. Scatterplot (Numerical-Numerical)

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='smoker', size='size') #Multivariate
```

2. BarPlot (Numerical-Categorical)

```python
titanic.head()
```

```python
sns.barplot(data=titanic, x='Pclass', y='Age')
```

```python
sns.barplot(data=titanic, x='Pclass', y='Fare', hue='Sex')
```

3. Boxplot (Numerical-Categorical)

```python
sns.boxplot(data=titanic, x='Sex', y='Age', hue="Survived")
```

4. Displot(Numerical-Categorical)

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist=False)
plt.legend()
```

5. HeatMap (Categorical-Categorical)

```python
titanic.head(3)
```

```python
# In every Pclass how many people died and how many survived 
pd.crosstab(titanic['Pclass'], titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']), cmap="coolwarm")
```

```python
(titanic.groupby('Embarked')['Survived'].mean()*100)
```

6.ClusterMap (Categorical-categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']), cmap="coolwarm")
```
