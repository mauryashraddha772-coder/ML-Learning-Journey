```python
def Convert_fractional_year(fractional_year): 
    year = int(fractional_year)  
    fraction = fractional_year - year 
    days = round(fraction*365.25)
    date = pd.Timestamp(year=year,month=1,day=1) + pd.Timedelta(days=days) 
    return date 
ds2['X1 transaction date'] = ds2['X1 transaction date'].apply(Convert_fractional_year)
ds2.head(20)
```

```python
Q1 = ds["Age"].quantile(0.25)
```

```python
Q3 = ds["Age"].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower_limit = Q1 - 1.5 * IQR
```

```python
upper_limit = Q3 + 1.5 * IQR
```

```python
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = ds[(ds["Age"] < lower_limit) | (ds["Age"] > upper_limit)]
print(outliers)
print("Number of outliers:", outliers.shape[0])
```

```python
correlation_matrix = data[numerical_columns].corr()
correlation_matrix
```

```python
plt.figure(figsize=(8,6)) 
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".3f", linewidths=0.5) 
plt.title("correlation matrix of numerical variables") 
plt.tight_layout()
plt.show()
```

```python
corr_pairs = correlation_matrix.where(
    ~np.eye(correlation_matrix.shape[0], dtype=bool)
)
corr_pairs
```

```python
highest_positive = corr_pairs.stack().idxmax() 
highest_positive_value = corr_pairs.stack().max()
```

```python
print("\n=======HIGHEST POSITIVE CORRELATION==========") 
print("Variables:", highest_positive)
print("Correlation coefficient:", round(highest_positive_value, 4))
```

```python
lowest_positive = corr_pairs.stack().idxmin()
lowest_positive_value = corr_pairs.stack().min() 
```

```python
print("\n=====Lowest Positive Correlation======") 
print("Variables:", lowest_positive)
print("Correlation Coefficient:", round(lowest_positive_value, 4))
```
