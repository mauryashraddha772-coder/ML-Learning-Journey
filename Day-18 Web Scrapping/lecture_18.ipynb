```python
import pandas as pd 
import requests
from bs4 import BeautifulSoup
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
```

```python
webpage =requests.get('https://www.ambitionbox.com/list-of-companies?campaign=homepage_companies_widget&page=1', headers=headers).text
```

```python
import lxml
```

```python
soup = BeautifulSoup(webpage, 'lxml')
```

For 1 company

we r going in every div this time

```python
company=soup.find_all('div',class_="companyCardWrapper")
```

```python
len(company)
```

```python
name=[]
rating=[]
reviews = []
information = []
for i in company:
    name.append(i.find('h2').text.strip()) #strip to remove extra charachter
    rating.append(i.find_all('div', class_="companyCardWrapper__companyRating")[0].text.strip())
    reviews.append(i.find('a', class_="companyCardWrapper__ActionWrapper").text.strip())
    information.append(i.find('div', class_="companyCardWrapper__tertiaryInformation").text.strip()) 
    
d = {'name': name, 
     'rating': rating, 
     'reviews': reviews, 
     'information': information} 

df =pd.DataFrame(d)
df
```
